```python
# The map: each room lists the rooms you can walk to next.
map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": [],           # G is the treasure! No need to go further.
}

print("Our map:")
print("            A  (Start)")
print("           / \\")
print("          B   C")
print("           \\ / ")
print("            D")
print("            |")
print("            G  (Treasure 💎)")
print()
print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal) :
    to_do = [start] # this is the to do (to visit nodes) after the first node 
    visited =[] # No of nodes visited
    order = [] # Order of bsf

    while to_do :
        room = to_do.pop(0)
        if room in visited:
            continue
        visited.append(room)
        order.append(room)

        if room == goal:
            return order
        
        for i in map_rooms[room]:
            to_do.append(i)
    return order

order = bfs("A", "G")
print("DFS checked the room is this order :", order)
print("No of rooms DFS checked : ", len(order))
            

```
