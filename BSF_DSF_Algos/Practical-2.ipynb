```python
# The map: each room lists the rooms you can walk to next.
map_rooms = {
    "A": ["B", "C"],   # from A you can go to B or C
    "B": ["D"],        # from B you can go to D
    "C": ["D"],        # from C you can go to D
    "D": ["G"],        # from D you can go to G
    "G": [],           # G is the treasure! No need to go further.
}

print("Our map:")
print("            A  (Start)")
print("           / \\")
print("          B   C")
print("           \\ / ")
print("            D")
print("            |")
print("            G  (Treasure 💎)")
print()
print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal) :
    to_do = [start] # this is the to do (to visit nodes) after the first node 
    visited =[] # No of nodes visited
    order = [] # Order of bsf

    while to_do :
        room = to_do.pop(0)
        if room in visited:
            continue
        visited.append(room)
        order.append(room)

        if room == goal:
            return order
        
        for i in map_rooms[room]:
            to_do.append(i)
    return order

order = bfs("A", "G")
print("BFS checked the room is this order :", order)
print("No of rooms DFS checked : ", len(order))
            

```

```python
def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()

        if room in visited:
            continue
        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order

order = dfs("A", "G")
print("DFS checked the rooms in this order:", order)
print("Number of rooms DFS checked:", len(order))
```

```python
door_cost = {
    ("A", "B") : 1,
    ("B", "D") : 1,
    ("A", "C") : 5,
    ("C", "D") : 1,
    ("D", "G") : 1
}
```

```python
def path_cost(path):
    total = 0
    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total += door_cost[door]
    return total

path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Path 1 A > B > D > G", path_cost(path1))
print("Path 2 A > C > D > G", path_cost(path2))
```

```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("Path 1 cost:", cost1)
print("Path 2 cost:", cost2)
print()
print("The cheapest path is:", best_path, "with cost", best_cost, "💰")

bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("RESULTS")
print("-" * 40)
print("BFS  (nearest first) checked:", len(bfs_order), "rooms ->", bfs_order)
print("DFS  (deep first)    checked:", len(dfs_order), "rooms ->", dfs_order)
print("Cheapest path:", best_path, "with cost", best_cost)
print("-" * 40)

print()
```

```python
door_cost[("A", "B")] = 3
door_cost[("A", "C")] = 7

cost1 = path_cost(path1)
cost2 = path_cost(path2)

print("Now path 1 (through B) costs : ", cost1)
print("Now path 2 (through C) costs : ", cost2)

if cost1 < cost2:
    print("Cheapest is Path 1 - A > B > D > G")
else :
    print("Cheapest is Path 2 - A > C > D > G")
```
