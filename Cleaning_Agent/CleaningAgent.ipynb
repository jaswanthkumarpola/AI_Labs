```python
room = ["D", "C", "C", "D", "C", "C"]
```

```python
def show_room(room):
    picture = ""
    for spot in room :
        if spot == "D":
            picture += "🙂 "
        else:
            picture += "👍🏻 "
    print(picture)        
print("Room now : ")
show_room(room)        
```

```python
# Creating the cleaning part. 
def clean(spot):
    if spot == "D":
        return "Cleaned spot"
    elif spot == "C":
        return "Already Clean"
    else :
        return 1 
# Testing 
result = clean("D")
print("After verifying", result)
```

```python
print("Before cleaning :")
show_room(room)
print()

for i in range(len(room)):
    room[i] = clean(room[i])
    print("Updating the spot clead virtually :", i+1, ":") 
    show_room(room)
 
print("After cleaning :")
show_room(room)
```

```python
room2 = ["D", "C", "D", "D", "C"]

cleaned = 0
for i in range(len(room2)):
    if room2[i] == "D":
        cleaned = cleaned + 1     # count this dirty spot
    room2[i] = clean(room2[i])

print("The robot cleaned", cleaned, "dirty spots. 🧹")
print("The room is now:")
show_room(room2)
```

```python
my_room = ["D", "D", "D", "D", "D"]

print("My room before:")
show_room(my_room)

for i in range(len(my_room)):
    my_room[i] = clean(my_room[i])

print("My room after the robot cleaned it:")
show_room(my_room)
print("All clean! 🌟")
```
