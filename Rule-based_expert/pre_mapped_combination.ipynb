```python
from typing import List, Tuple, Dict

RECOMMENDATIONS: Dict[str, Dict[str, Tuple[str, str]]]= {
    "relxed":{
        "snack": ("comedy", "popcorn & soda"),
        "heavy": ("Slice of Life", "Pizza & cold drink")
    },
    "energetic":{
        "snack": ("Action", "Nachos & Cheese Dip"),
        "heavy": ("Sci-fi", "Burger & Fries"),
    },
    "thoughtful":{
        "snack": ("Mystery/Thriller", "Pretzels & Coffe"),
        "heavy": ("Drama/Documentry", "Pasta & Garlic Bread"),
    }
}
```

```python
def get_choice(prompt: str, options: List[str]) -> str:
    """Helper function to validate simple text options."""
    while True:
        user_input = input(f"{prompt} ({'/'.join(options)}): ").strip().lower()
        if user_input in options:
            return user_input
        print(f"Invalid choice! Please pick one of: {', '.join(options)}")


def main():
    print("==================================================")
    print("      MOVIE & MEAL RECOMMENDATION SYSTEM")
    print("==================================================")
    print("Answer 2 quick questions to get your evening recommendation!\n")

    # Step 1: Collect User Inputs
    mood = get_choice("What is your current mood?", ["relaxed", "energetic", "thoughtful"])
    meal_type = get_choice("What kind of food do you prefer right now?", ["snack", "heavy"])

    # Step 2: Apply Rules / Look Up
    movie, meal = RECOMMENDATIONS[mood][meal_type]

    # Step 3: Display Output
    print("\n--------------------------------------------------")
    print("  YOUR MATCHED RECOMMENDATION")
    print("--------------------------------------------------")
    print(f"  Recommended Movie Genre : {movie}")
    print(f"  Recommended Food Combo   : {meal}")
    print("--------------------------------------------------")


main()
```
