```python
def troubleshooting(performance_slow, network_slow, power_on):
    if performance_slow == 'yes':
        return "Close unused programs and files form task manager"
    elif network_slow == "yes":
        return "Try to restart the router/reconnect try to the network"
    elif power_on == "yes":
        return "Problem in laptop/try to hard restart"
    else:
        return "No problems found, everything is fine👍🏻"
```

```python
performance_slow = input("Are you experincing performance lag?: ").lower()
network_slow = input("Is your network slow?: ").lower()
power_on = input("Can't see the screen on?: ").lower()

advice = troubleshooting(performance_slow, network_slow, power_on)
print("Expert System Advice is :")
print(advice)
```
