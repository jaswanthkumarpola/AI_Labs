```python
import pandas as p
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
df = p.read_csv("student_course_reviews_500.csv")
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
from sklearn.tree import DecisionTreeClassifier

train_x = df["review_text"]
train_y = df["sentiment"]

x_train, x_test, y_train, y_test = train_test_split(
    train_x, train_y, test_size=0.2, random_state=42, stratify=train_y
)

vectorizer = CountVectorizer()
x_train_vec = vectorizer.fit_transform(x_train)
x_test_vec = vectorizer.transform(x_test)

model = DecisionTreeClassifier(
    random_state=42,
    class_weight="balanced"
)
model.fit(x_train_vec, y_train)

predictions = model.predict(x_test_vec)

print("Decision Tree Classification Report:")
print(classification_report(y_test, predictions))

nb_model = MultinomialNB()
nb_model.fit(x_train_vec, y_train)

nb_predictions = nb_model.predict(x_test_vec)

print("Naive Bayes Classification Report:")
print(classification_report(y_test, nb_predictions))
```

```python
new_reviews = [
    "The instructor was knowledgeable, and the lessons were clear and engaging.",
    "The course was poorly organized, and the assignments were confusing.",
    "The workload was reasonable, and the course content was satisfactory.",
    "Excellent practical activities and helpful feedback from the instructor."
]

new_reviews_vec = vectorizer.transform(new_reviews)

decision_tree_predictions = model.predict(new_reviews_vec)
naive_bayes_predictions = nb_model.predict(new_reviews_vec)

new_data = p.DataFrame({
    "review_text": new_reviews,
    "decision_tree_prediction": decision_tree_predictions,
    "naive_bayes_prediction": naive_bayes_predictions
})

new_data
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

rf_model.fit(x_train_vec, y_train)
rf_predictions = rf_model.predict(x_test_vec)

print("Random Forest Classification Report:")
print(classification_report(y_test, rf_predictions))
```

```python
new_reviews = [
    "The instructor was knowledgeable, and the lessons were clear and engaging.",
    "The course was poorly organized, and the assignments were confusing.",
    "The workload was reasonable, and the course content was satisfactory.",
    "Excellent practical activities and helpful feedback from the instructor."
]

new_reviews_vec = vectorizer.transform(new_reviews)

decision_tree_predictions = model.predict(new_reviews_vec)
naive_bayes_predictions = nb_model.predict(new_reviews_vec)
rf_predictions = rf_model.predict(new_reviews_vec)


new_data = p.DataFrame({
    "review_text": new_reviews,
    "decision_tree_prediction": decision_tree_predictions,
    "naive_bayes_prediction": naive_bayes_predictions,
    "randomforest_prediction": rf_predictions
})

new_data
```
