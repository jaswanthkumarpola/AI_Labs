```python
intial_state = {"At_A"}
goal = "At_C"
```

```python
actions = [
    {
        "name": "Move A to B",
        "pre": {"At_A"},
        "Add": {"At_B"},
        "delete": {"At_C"}
    },
    {
        "name": "Move B to C",
        "pre": {"At_B"},
        "Add": {"At_C"},
        "delete": {"At_B"}

    }
]
```

```python
def find_action(goal):
    for action in actions:
        if goal in action["Add"]:
            return action
    return None


def make_plan(goal, state, plan):
    if goal in state:
        return True

    action = find_action(goal)
    if action is None:
        return False

    for condition in action["pre"]:
        if not make_plan(condition, state, plan):
            return False

    state.difference_update(action["delete"])
    state.update(action["Add"])
    plan.append(action["name"])

    return True


state = intial_state.copy()
plan = []

make_plan(goal, state, plan)

print("plan:")
for step in plan:
    print(step)
```
